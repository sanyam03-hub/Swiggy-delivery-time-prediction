# Swiggy Delivery Time Prediction - Exploratory Data Analysis

This notebook explores the synthetic Swiggy delivery dataset and provides insights into the factors affecting delivery time.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
from pathlib import Path

# Set style
plt.style.use('seaborn-v0_8')
sns.set_palette('husl')

# Load data
data_path = '../data/synthetic/swiggy_delivery_data.csv'
df = pd.read_csv(data_path)

print(f"Dataset shape: {df.shape}")
print(f"\nColumn info:")
df.info()
```

## 1. Basic Statistics

```python
# Basic statistics
df.describe()
```

## 2. Delivery Time Analysis

```python
# Delivery time distribution
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# Histogram
axes[0,0].hist(df['delivery_time_minutes'], bins=50, alpha=0.7)
axes[0,0].set_title('Distribution of Delivery Times')
axes[0,0].set_xlabel('Delivery Time (minutes)')

# Box plot by city
df.boxplot('delivery_time_minutes', by='city', ax=axes[0,1])
axes[0,1].set_title('Delivery Time by City')

# Box plot by weather
df.boxplot('delivery_time_minutes', by='weather_condition', ax=axes[1,0])
axes[1,0].set_title('Delivery Time by Weather')

# Box plot by food category
df.boxplot('delivery_time_minutes', by='food_category', ax=axes[1,1])
axes[1,1].set_title('Delivery Time by Food Category')
axes[1,1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

## 3. Time-based Patterns

```python
# Time-based analysis
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Average delivery time by hour
hourly_avg = df.groupby('order_hour')['delivery_time_minutes'].agg(['mean', 'count']).reset_index()
axes[0].plot(hourly_avg['order_hour'], hourly_avg['mean'], marker='o')
axes[0].set_title('Average Delivery Time by Hour')
axes[0].set_xlabel('Hour of Day')
axes[0].set_ylabel('Average Delivery Time (minutes)')
axes[0].grid(True)

# Orders by day of week
day_names = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
daily_avg = df.groupby('day_of_week')['delivery_time_minutes'].agg(['mean', 'count']).reset_index()
daily_avg['day_name'] = [day_names[i] for i in daily_avg['day_of_week']]

axes[1].bar(daily_avg['day_name'], daily_avg['mean'])
axes[1].set_title('Average Delivery Time by Day')
axes[1].set_ylabel('Average Delivery Time (minutes)')
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

## 4. Geographic Analysis

```python
# Distance analysis
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Distance distribution
axes[0].hist(df['distance_km'], bins=30, alpha=0.7)
axes[0].set_title('Distribution of Delivery Distances')
axes[0].set_xlabel('Distance (km)')

# Distance vs delivery time
axes[1].scatter(df['distance_km'], df['delivery_time_minutes'], alpha=0.5)
axes[1].set_title('Distance vs Delivery Time')
axes[1].set_xlabel('Distance (km)')
axes[1].set_ylabel('Delivery Time (minutes)')

# Traffic vs delivery time
axes[2].scatter(df['traffic_density'], df['delivery_time_minutes'], alpha=0.5)
axes[2].set_title('Traffic Density vs Delivery Time')
axes[2].set_xlabel('Traffic Density')
axes[2].set_ylabel('Delivery Time (minutes)')

plt.tight_layout()
plt.show()
```

## 5. Correlation Analysis

```python
# Correlation matrix
numeric_cols = df.select_dtypes(include=[np.number]).columns
correlation_matrix = df[numeric_cols].corr()

plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0)
plt.title('Correlation Matrix of Numeric Features')
plt.tight_layout()
plt.show()
```

## 6. Key Insights

Based on the exploratory data analysis:

1. **Distance Impact**: There's a strong positive correlation between distance and delivery time
2. **Time Patterns**: Peak hours (lunch and dinner) show higher average delivery times
3. **Weather Effect**: Bad weather conditions increase delivery times
4. **Traffic Impact**: Higher traffic density correlates with longer delivery times
5. **Restaurant Efficiency**: Higher-rated restaurants tend to have shorter delivery times
6. **City Differences**: Different cities show varying delivery time patterns

These insights validate our feature engineering approach and confirm that our synthetic data captures realistic delivery patterns.
